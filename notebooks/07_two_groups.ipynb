# Two groups of dots: averaging or choosing?

Two groups of dots move side by side at a separation L, in the paradigm of Sridhar et al. (2021): does a simulated
fish swim between them (averaging) or follow one (choosing)? M6 and F at both seeds, under four rules for combining
the two groups, in a dish of twice the radius, with the fish placed equidistant from the two groups at onset. The
simulations are cached in `output/two_targets/`; `uv run python -m twin.two_targets run-report` redoes them
(about 2 CPU-hours) and `uv run python -m twin.two_targets validate` checks that one group reproduces the standard
simulator.

## The bifurcation maps (Figures fig:bifurcation and fig:app-bif-*)
Columns: the rules (effects = the groups' effects summed; add; avg; nearest group only) and the two references,
only group A shown and no group shown. Each column of a map is the distribution of the fish's lateral position at
one separation, over the bouts behind the pair, normalised to its maximum; dashed: the two groups' tracks.
20-60 s after onset, then the whole block.
